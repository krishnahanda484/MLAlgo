# AgglomerativeClustering Hierarchical

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv(r"C:\Users\conte\Downloads\archive (5)\Iris.csv")
```

```python
dataset.head(5)
```

```python
dataset = dataset.iloc[:,:-1]
dataset.head(5)
```

```python
sns.pairplot(data=dataset)
plt.show()
```

```python
import scipy.cluster.hierarchy as sc
```

```python
plt.figure(figsize=(20,10))
sc.dendrogram(sc.linkage(dataset,method='single',metric='euclidean'))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
ac = AgglomerativeClustering(n_clusters=2,linkage='single')
dataset["Predict"] = ac.fit_predict(dataset)
```

```python
sns.pairplot(data=dataset,hue="Predict")
plt.show()
```
